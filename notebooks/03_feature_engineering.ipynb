# Developer Career Intelligence Platform - 03 Feature Engineering & Analytical Scoring

This notebook covers:
1. Multi-dimensional Developer Intelligence Score calculation.
2. Developer Growth Velocity (DGV) time-series evaluation.
3. Skill Momentum classifications (Rising, Stable, Declining, New, Dormant).
4. Project Complexity Score and Portfolio Quality Auditing.

```python
import sys
sys.path.append('..')
import pandas as pd
from database.database import DatabaseManager
from analytics.developer_score import DeveloperIntelligenceScorer
from analytics.growth_velocity import GrowthVelocityAnalyzer
from analytics.skill_momentum import SkillMomentumAnalyzer
from analytics.portfolio_auditor import PortfolioAuditor

db = DatabaseManager('../data/developer_intelligence.db')
user = db.get_user_by_username('alex-datascientist')
repos = db.get_repositories(user['id'])
langs = db.get_languages(user['id'])
commits = db.get_commits(user['id'])
prs = db.execute_query('SELECT * FROM pull_requests WHERE user_id = ?', (user['id'],))

# Calculate Scores
scores = DeveloperIntelligenceScorer.compute_scores(user, repos, langs, commits, prs)
print("Developer Intelligence Score:", scores['overall_score'])
pd.DataFrame([{
    'Overall': scores['overall_score'],
    'Depth': scores['technical_depth'],
    'Breadth': scores['technical_breadth'],
    'Consistency': scores['consistency'],
    'Complexity': scores['project_complexity'],
    'Collaboration': scores['collaboration'],
    'Adaptability': scores['adaptability']
}])
```

```python
# Developer Growth Velocity (DGV)
growth = GrowthVelocityAnalyzer.analyze_growth(commits, repos, langs)
print("Growth Velocity Score:", growth['growth_velocity_score'])
print("Velocity Tier:", growth['velocity_tier'])
pd.DataFrame(growth['annual_trajectories'])
```

```python
# Skill Momentum Tracking
momentum = SkillMomentumAnalyzer.compute_momentum(langs, repos, commits)
pd.DataFrame(momentum)[['technology', 'momentum', 'total_bytes', 'reason']]
```
